# NCAA tournament forecasting — public evidence

Aggregate evidence and negative findings. This notebook renders disclosed result tables only; it is not the private training implementation. No executed outputs are supplied in this download.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.io as pio
from IPython.display import display
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT=ROOT.parent
DATA=pd.read_csv(ROOT/'data/validation_metrics.csv',dtype={'round':str})
pio.renderers.default='plotly_mimetype'
display(DATA.head())
```

## Within-experiment comparisons
Arm names are deliberately generic. Do not compare different population/season rows as though they are a single leaderboard.

```python
ROUND='19'
view=DATA.loc[DATA['round'].eq(ROUND)]
fig=px.line(view,x='season',y='brier',color='arm',markers=True,title='Round '+ROUND+' — exploratory historical Brier',labels={'brier':'Brier (lower is better)'})
fig.show()
```

## Evidence coverage
Counts measure reported comparisons, not novel features or independent tests.

```python
coverage=DATA.groupby(['round','population']).size().rename('reported_rows').reset_index()
px.bar(coverage,x='round',y='reported_rows',color='population',title='Disclosed experiment coverage').show()
```
